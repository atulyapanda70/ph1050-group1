```python
data=pd.read_csv("planets_300.csv")
df=data[['planet_id','a_AU','M_star_solar','T_years',]]
print(df.head())
```

```python
df['T**2*M/a**3']=(df['T_years']**2)*(df['M_star_solar'])/(df['a_AU']**3)
print(df[['planet_id','T**2*M/a**3']].head())
```

```python
values=df['T**2*M/a**3'].to_numpy()
print("Count:",np.size(values))
print("Mean:",np.mean(values))
print("Median:",np.median(values))
print("Standard Deviation:",np.std(values))
print("Minimum:",np.min(values))
print("Maximim:",np.max(values))
```

```python

plt.hist(values,bins=30)
plt.xlabel("T**2*M/a**3")
plt.ylabel("Number Of Planets")
plt.title("Distribution of T**2*M/a**3")
plt.show()
```

```python
df['logT']=np.log10(df['T_years'])
df['loga']=np.log10(df['a_AU'])
print(df[['planet_id','logT','loga']].head())
```

```python
plt.scatter(df['loga'], df['logT'], s=8, alpha=0.5)
plt.xlabel("loga")
plt.ylabel("logT")
plt.title("Loga vs LogT")
plt.show()
```

```python
slope,intercept=np.polyfit(df['loga'],df['logT'],1)
print("Slope:",slope)
print("Intercept:",intercept)
```
