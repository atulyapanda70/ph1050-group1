```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

Genrating random orbital radii and stellar masses

```python
# purpose- gotta make synthetic data for 300 planets, semi major axis between 0.05 to 30 AU, stellar mass between 0.5 to 1.5 solar masses, and noise 3%



rng = np.random.default_rng(42)  #Creates a random number generator, 42 is just a seed

n = 300

a = rng.uniform(0.05, 30, n) #Generates 300 random numbers between 0.05 and 30 (in AU)

M = rng.uniform(0.5, 1.5, n) # Generates 300 random numbers between 0.5 and 1.5 (in solar masses)

T_true = np.sqrt(a**3 / M) #contains the theoretical orbital period of every generated planet, measured in Earth years

#Real astronomical observations are not perfectly accurate. introduce 3% noise.We'll use multiplicative Gaussian noise

noise = rng.normal(0, 0.03, n)

T_observed = T_true * (1 + noise)



# create the dataset
df = pd.DataFrame({
    "planet_id": [f"P{i:03d}" for i in range(1, n+1)],
    "a_AU": a,
    "M_star_solar": M,
    "T_true_years": T_true,
    "noise_fraction": noise,
    "T_years": T_observed
})

df.to_csv("planets_300.csv", index=False)

print(df.head())
print("Number of planets:", len(df))

```
